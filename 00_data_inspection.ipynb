```python
import pandas as pd
import os

# 1. 设置路径
BASE_PATH = 'raw data'
files = [
    'train.csv', 
    'test.csv', 
    'stores.csv', 
    'oil.csv', 
    'holidays_events.csv', 
    'transactions.csv'
]

# 2. 定义一个“体检医生”函数
def inspect_dataset(file_name):
    file_path = os.path.join(BASE_PATH, file_name)
    df = pd.read_csv(file_path)
    
    print(f"\n{'='*20} 📂 正在检查: {file_name} {'='*20}")
    
    # 1. 规模 (Shape)
    print(f"1️⃣  数据规模 (行, 列): {df.shape}")
    
    # 2. 缺失值 (Missing Values)
    missing = df.isnull().sum()
    missing = missing[missing > 0] # 只显示有缺失的列
    if not missing.empty:
        print(f"2️⃣  ⚠️ 发现缺失值:\n{missing}")
    else:
        print("2️⃣  ✅ 没有发现缺失值")
        
    # 3. 数据类型概览
    print("\n3️⃣  数据列名与类型:")
    print(df.dtypes)
    
    # 4. 预览 (Head)
    print(f"\n4️⃣  前 3 行预览:")
    display(df.head(3)) # Jupyter里用display显示表格更好看，print也行

# 3. 挨个检查
for file in files:
    inspect_dataset(file)
```
